### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here\
import math

x1 = float(input("Enter the x1 cordinate: "))
y1 = float(input("Enter the y1 cordinate: "))
x2 = float(input("Enter the x2 cordinate: "))
y2 = float(input("Enter the y2 cordinate: "))

result = math.sqrt((x2-x1)**2 + (y2-y1)**2)
print(result)
````

For example:
Input:

Output:

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here

n = int(input("Enter a n number"))

sum_sqr = 0
for i in range (1,n+1):
  sum_sqr += i**2
  print(sum_sqr)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here

first = int(input("Enter the first term: "))
second = int(input("Enter a second term: "))


n = int(input("Enter the Nth value"))
d = second - first
nth_term = first + (n-1) * d
print("Nth term is: ", nth_term)


````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
import math
tank_height = float(input("Enter height of tank"))
tank_width = float(input("Enter width of tank"))
tank_breadth = float(input("Enter breadth of tank"))

glass_height = float(input("Enter height of glass"))
glass_radius = float(input("Enter radius of glass"))

tank_volume = tank_height * tank_width * tank_breadth
glass_volume = math.pi * glass_height * (glass_radius ** 2)

glasses = tank_volume / glass_volume
print("Number of glasses = " , int(glasses))



````
