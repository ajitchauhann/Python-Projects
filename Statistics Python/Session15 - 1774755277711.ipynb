```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

```python
df = pd.read_csv('cust_seg.csv')
df.head()
```

```python
# Q1. 1 - sample t-test
pop_mean = 50

# H0: sample mean <= 50
# Ha: sample mean > 50
```

```python
alpha = 0.05
stat, p_value = stats.ttest_1samp(df.Post_usage_1month, pop_mean, alternative = 'greater')
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; Card usage has improved significantly.')
else:
    print('Fail to reject the null hypothesis; Card usage remained same or dropped')
```

```python
# Q2. Relational/Paired t-test

var1 = 'pre_usage'
var2 = 'Post_usage_1month'

# H0: mean_var2 <= mean_var1
# Ha: mean_var2 > mean_var1
```

```python
alpha = 0.05
stat, p_value = stats.ttest_rel(df[var1], df[var2], alternative='less')
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; Last campaign was successful.')
else:
    print('Fail to reject the null hypothesis; Last campaign was not successful')
```

```python
# Q3. Independent sample t-test

# H0: mean_male = mean_female
# Ha: mean_male != mean_female
```

```python
alpha = 0.05
sample_male = df.loc[df.gender == 0, 'pre_usage']
sample_female = df.loc[df.gender == 1, 'pre_usage']
```

```python
stat, p_value = stats.ttest_ind(sample_male, sample_female)
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; The spending behavior is different')
else:
    print('Fail to reject the null hypothesis; The spending behavior is same')
```

```python
# Q4.  ANOVA test (Analysis of variance)

# H0: Samples are from same population i.e., mean of samples are equal
# Ha: Samples are from different population i.e., mean of samples are unequal
```

```python
alpha = 0.05
s1 = df.loc[df.segment == 1, 'pre_usage']
s2 = df.loc[df.segment == 2, 'pre_usage']
s3 = df.loc[df.segment == 3, 'pre_usage']
```

```python
stat, p_value = stats.f_oneway(s1, s2, s3)
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; Atleast one segment is from different population and therefore, customers are spending at different levels')
else:
    print('Fail to reject the null hypothesis; The segments are from the same populations and therefore, the spending behavior is same')
```

```python
# Q5. Chi-square test

# H0: There is no relationship between region and segment
# Ha: There is relationship between region and segment

alpha = 0.05
cross_table = pd.crosstab(df.region, df.segment)
```

```python
stat, p_value, _, _ = stats.chi2_contingency(cross_table)
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; There is a relationship between segment and region')
else:
    print('Fail to reject the null hypothesis; There is no relationship between segment and region')
```

```python
# Q6. Correlation test

# H0: There is no relationship between latest month usage and pre_usage
# Ha: There is a relationship between latest month usage and pre_usage
```

```python
alpha = 0.05
stat, p_value = stats.pearsonr(df.Latest_mon_usage, df.pre_usage)
```

```python
if p_value <= alpha:
    print('Reject the null hypothesis; There is a relationship between latest month usage and pre_usage')
else:
    print('Fail to reject the null hypothesis; There is no relationship between latest month usage and pre_usage')
```
