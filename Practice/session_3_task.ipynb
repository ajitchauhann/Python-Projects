```python
# Age group Categorization
# Q1. Classify a person's age group: Child (< 13), Teenager(13,19),Adult(20-59),Senior(60+).

age_group = int(input("Enter a persons age group"))

if age_group <=13:
   print('Child')
elif age_group <= 20:
   print('Teenager')
elif age_group <=60:
   print('Adult')
else:
   print('Senior')
```

```python
# Grade calculator
# Que3:- Assign a letter grade based on a student's score: A(90-100),B(80-89),C(70-79),D(60-69),E(below 60).



score = 90

if score >= 101 or score < 0:
  print("Please verify your grade")

else:

   if score >= 90:
    grade = 'A'
   elif score >= 80:
    grade = 'B'
   elif score >= 70:
    grade = 'C'
   elif score >= 60:
    grade = 'D'
   else:
    grade = 'E'

   print("Grade is: ",grade)
```

```python
# Password Strength Checker
# Que 8:- Check if a password is "Weak", "Medium", or "Strong". Criteria: < 6 chars (Weak), 6-10 chars (Medium), >10 chars (Strong).

password = "Sec"

if len(password) < 6:
  strength = "Weak"
elif len(password) <= 10:
  strength = "Medium"
else:
  strength = "Strong"

print("Password strength is: ", strength)
```

```python
# Leap Year Checker
# Que 9:- Determine if a year is a leap year. (Leap years are divisible by 4, but not by 100 unless also divisible by 400).

year = 2024

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
  print(year, " is a leap year")
else:
  print(year, "is NOT a leap year")
```
